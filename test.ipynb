```python
class Parent:
    def __init__(self, name):
        self.name = name

class Child(Parent):
    @classmethod
    def from_string(cls, name):
        return cls(name)
```

```python
p = Parent('Vitia')
type(p)
```

```python
pp = Child('Petia')
type(pp)
```

```python
ppp = Child.from_string('Vania')
type(ppp)
```
