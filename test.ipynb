```python
def w1(x_em, r_em, r, Eph):
    """
    pair production probability
    Eph should be normalized to m_e c^2
    return: probability density (in w dl, dl expressed in R star)
    """
    if (r - r_em) / r_em < 1e-15:
        return 0
    psi_inf = r_em / (4/3 * 1200000.0 / 21667 / (x_em) * np.sqrt(r_em))
    # psi = self.psi_inf(x_em, r_em) * (1 - r_em/r)
    psi = psi_inf * (1 - r_em/r)
    return 0.23 * 1 / 137 / (3.86e-11/1200000) * 1/r**3 / (44.152) * psi * np.exp(-8/3 / Eph * 44.152/ (1/r**3) / psi)
```
